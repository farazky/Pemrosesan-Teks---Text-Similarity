Pecah jadi kalimat

```python
sentences_raw = re.split(r'(?<=[.!?])\s+', text.strip())
sentences_raw = [s for s in sentences_raw if len(s.strip()) > 20]
sentences_raw = list(dict.fromkeys(sentences_raw))   # buang kalimat duplikat (teksmu tertempel 2x)

print("Jumlah kalimat:", len(sentences_raw))
```

```python
def preprocess(text):
    text = text.lower()
    text = text.translate(str.maketrans('', '', string.punctuation))
    text = re.sub(r'\s+', ' ', text).strip()
    text = stemmer.stem_kalimat(text)
    text = " ".join([w for w in text.split() if w not in sw])
    return text

docs = [preprocess(s) for s in sentences_raw]

# jaga agar docs dan kalimat asli tetap sejajar
pairs = [(s, d) for s, d in zip(sentences_raw, docs) if d.strip() != '']
sentences_raw = [p[0] for p in pairs]
docs = [p[1] for p in pairs]

print("Jumlah dokumen:", len(docs))
print(docs[:3])
```

**Cosine Similarity**

vektorisasi untuk cosine (BoW dan TF-IDF)

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

X_bow   = CountVectorizer().fit_transform(docs)
X_tfidf = TfidfVectorizer().fit_transform(docs)

print("Shape BoW   :", X_bow.shape)
print("Shape TF-IDF:", X_tfidf.shape)
```

Hitung Cosine

```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

cos_bow   = cosine_similarity(X_bow)
cos_tfidf = cosine_similarity(X_tfidf)

iu = np.triu_indices(len(docs), 1)   # pasangan unik, tanpa diagonal
print("Rata-rata cosine BoW   :", cos_bow[iu].mean())
print("Rata-rata cosine TF-IDF:", cos_tfidf[iu].mean())
```

Pasangan kalimat paling mirip (cosine)

```python
def top_pairs(M, name, n=3):
    print(f"\n=== {name} ===")
    vals = M[iu]
    for k in np.argsort(vals)[::-1][:n]:
        i, j = iu[0][k], iu[1][k]
        print(f"{vals[k]:.3f} | K{i}: {sentences_raw[i][:70]}")
        print(f"        | K{j}: {sentences_raw[j][:70]}")

top_pairs(cos_bow, "Cosine BoW")
top_pairs(cos_tfidf, "Cosine TF-IDF")
```

**JACCARD SIMILARITY**

Vektorisasi untuk Jaccard

```python
X_ohe = CountVectorizer(binary=True).fit_transform(docs)
print("Shape One-Hot:", X_ohe.shape)
```

Hitung Jaccard

```python
from scipy.spatial.distance import pdist, squareform

jac = 1 - squareform(pdist(X_ohe.toarray(), metric='jaccard'))   # similarity = 1 - distance
print("Rata-rata Jaccard:", jac[iu].mean())
```

Pasangan kalimat paling mirip (Jaccard)

```python
top_pairs(jac, "Jaccard")
```
